```python
from decimal import Decimal, ROUND_HALF_UP

def denormalize_exact(x_norm_str, min_str, max_str, decimals=2):
    x_norm = Decimal(str(x_norm_str))
    min_val = Decimal(str(min_str))
    max_val = Decimal(str(max_str))
    
    # Exact base-10 math
    x_denorm = x_norm * (max_val - min_val) + min_val
    
    # Quantize to target precision
    quantizer = Decimal('1.' + '0' * decimals) if decimals > 0 else Decimal('1')
    return x_denorm.quantize(quantizer, rounding=ROUND_HALF_UP)
```

```python

def preprocess_missing_data(df: pd.DataFrame) -> pd.DataFrame:
    """
    Performs missing data reconstruction, outlier handling (4-sigma), 
    and climatological imputation according to the paper methodology.
    """
    # Create a copy to prevent modifying the original dataframe
    processed_df = df.copy()
    # processed_df['timestamp'] = pd.to_datetime(processed_df['timestamp'])
    
    # Define features to process
    sensor_features = [
        'temperature', 'irradiance', 'pressure', 'windspeed', 
        'humidity', 'dewpoint', 'windbearing', 'cloudcover'
    ]
    target_feature = 'kwh'

    # --- STEP 1: TEMPORAL GRID RECONSTRUCTION ---
    reindexed_slices = []
    
    # Process each site separately to handle different operational timeframes
    for site_id, site_data in processed_df.groupby('site_id'):
        min_date = site_data['timestamp'].min().date()
        max_date = site_data['timestamp'].max().date()
        
        # Generate complete sequence of calendar days
        all_days = pd.date_range(start=min_date, end=max_date, freq='D')
        
        # Generate expected 19 hours (02:00 to 20:00) per day
        expected_grid = []
        for day in all_days:
            for hour in range(2, 21):  # 2 to 20 inclusive
                expected_grid.append(pd.Timestamp(day).replace(hour=hour))
                
        grid_df = pd.DataFrame({'timestamp': expected_grid})
        grid_df['site_id'] = site_id
        
        # Left merge original data onto the complete grid (missing hours become NaN)
        merged = pd.merge(grid_df, site_data, on=['site_id', 'timestamp'], how='left')
        
        # # Forward fill static identifiers (model_id, brand_id)
        # merged['model_id'] = merged['model_id'].ffill().bfill().astype(int)
        # merged['brand_id'] = merged['brand_id'].ffill().bfill().astype(int)
        
        reindexed_slices.append(merged)
        
    df_grid = pd.concat(reindexed_slices, ignore_index=True)

    # # --- STEP 2: OUTLIER DETECTION (4-SIGMA RULE) ---
    # for col in sensor_features:
    #     mean_val = df_grid[col].mean()
    #     std_val = df_grid[col].std()
        
    #     # Define relaxed 4-sigma boundaries
    #     lower_bound = mean_val - 4 * std_val
    #     upper_bound = mean_val + 4 * std_val
        
    #     # Convert outliers to NaN
    #     outliers_mask = (df_grid[col] < lower_bound) | (df_grid[col] > upper_bound)
    #     df_grid.loc[outliers_mask, col] = np.nan

    # --- STEP 3: HIERARCHICAL CLIMATOLOGICAL IMPUTATION ---
    # Extract helper temporal features
    df_grid['year'] = df_grid['timestamp'].dt.year
    df_grid['month'] = df_grid['timestamp'].dt.month
    df_grid['hour'] = df_grid['timestamp'].dt.hour

    # 1. Weather Features: Impute using Site-Year-Month-Hour group averages
    grp_year_month_hour = ['site_id', 'year', 'month', 'hour']
    climatology_year = df_grid.groupby(grp_year_month_hour)[sensor_features].transform('mean')
    for col in sensor_features:
        df_grid[col] = df_grid[col].fillna(climatology_year[col])
        
    # 2. Weather Features Fallback: Impute any remaining NaNs using Site-Month-Hour averages
    grp_month_hour = ['site_id', 'month', 'hour']
    climatology_global = df_grid.groupby(grp_month_hour)[sensor_features].transform('mean')
    for col in sensor_features:
        df_grid[col] = df_grid[col].fillna(climatology_global[col])

    # 3. Target Feature (kWh): Impute using Site-Month-Hour averages (as per paper Section A)
    target_climatology = df_grid.groupby(grp_month_hour)[target_feature].transform('mean')
    df_grid[target_feature] = df_grid[target_feature].fillna(target_climatology)

    # Clean up temporal helpers and return
    df_grid = df_grid.drop(columns=['year', 'month', 'hour'])
    
    # # Interpolate astronomical parameters if missing
    # if 'altitude' in df_grid.columns and 'azimuth' in df_grid.columns:
    #     df_grid['altitude'] = df_grid['altitude'].interpolate(method='linear')
    #     df_grid['azimuth'] = df_grid['azimuth'].interpolate(method='linear')
        
    return df_grid
```
